```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split

# Summarise tasks per project
task_summary = full_data.groupby('Project_ID').agg(
    Total_Planned_Hours=('Planned_Hours', 'sum'),
    Total_Actual_Hours=('Actual_Hours', 'sum'),
    Task_Count=('Task_ID', 'count')
).reset_index()

# Merge with projects table
project_data = task_summary.merge(projects, on='Project_ID', how='inner')

# --- Clean Budget: keep only digits and decimal point ---
project_data['Budget'] = project_data['Budget'].str.replace(r'[^\d.]', '', regex=True).astype(float)

# Create 'Delayed' column (1 = delayed, 0 = on time)
project_data['Delayed'] = (
    project_data['Total_Actual_Hours'] > project_data['Total_Planned_Hours']
).astype(int)

# Features and target
features = ['Task_Count', 'Budget', 'Department']
X = project_data[features].copy()
y = project_data['Delayed']

# One-hot encode Department
X_encoded = pd.get_dummies(X, columns=['Department'], drop_first=True)

# Split data
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

# Train model
model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

# Predict and evaluate
y_pred = model.predict(X_test)
accuracy = (y_pred == y_test).mean()
print(f"Model Accuracy: {accuracy:.2f}")
```

```python
# ---- Interactive Delay Predictor ----
print("🔮 Project Delay Predictor")
print("Enter the details of a new project to see if it's likely to be delayed.\n")

# Get user input
task_count = int(input("Number of tasks: "))
budget = float(input("Budget (in dollars, e.g. 200000): "))
department = input("Department (e.g. IT, Marketing, HR, Operations, Sales): ").strip()

# Create a one-row dataframe
new_project = pd.DataFrame({
    'Task_Count': [task_count],
    'Budget': [budget],
    'Department': [department]
})

# One-hot encode, aligning with training columns
new_encoded = pd.get_dummies(new_project, columns=['Department'], drop_first=True)
for col in X_encoded.columns:
    if col not in new_encoded.columns:
        new_encoded[col] = 0
new_encoded = new_encoded[X_encoded.columns]

# Predict
prediction = model.predict(new_encoded)[0]
probability = model.predict_proba(new_encoded)[0][1]  # probability of being delayed

# Output result
print("\n--- Prediction ---")
if prediction == 1:
    print(f"⚠️  This project is likely to be DELAYED (probability: {probability:.0%})")
else:
    print(f"✅  This project is likely to be ON TIME (probability of delay: {probability:.0%})")
```
